```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv(r"C:\Users\ANURAG DAS\Downloads\customer_churn_raw_dataset.csv",encoding="latin1")
df.info()
df.head()
for i in df.columns:
  print(df[i].value_counts())

```

```python
df.drop_duplicates(subset=['customer_id'],inplace=True)
df['customer_id'].value_counts()
avg_age=np.mean(df['age'])
df['age']=df['age'].fillna(avg_age)
df['age']=df['age'].astype(int)
df.info()
df.head(5)
```

```python
mod_location=df['location'].mode()[0]
print(mod_location)
df['location']=df['location'].fillna(mod_location)
mean_engagement=np.mean(df['engagement_score'])
df['engagement_score']=df['engagement_score'].fillna(mean_engagement)
df.info()
df.head()
```

```python
df['gender']=df['gender'].str.strip().str.title()
mapping={
    'M':'Male',
    'F':'Female',
    'Non-Binary':'Other'
}
df['gender']=df['gender'].replace(mapping)
df['gender'].value_counts()
```

```python
churn_rate=df['is_churned'].mean()*100
print(churn_rate)
type_wise_churn=df.groupby('subscription_type')['is_churned'].mean()*100.
print(type_wise_churn.round(2))
df.to_csv("customer_churn_cleaned.csv", index=False)
```
